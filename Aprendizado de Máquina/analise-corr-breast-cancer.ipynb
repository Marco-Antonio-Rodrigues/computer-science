Análise de correlações das features do dataset 'breast cancer'

```python
import pandas as pd
from sklearn.datasets import load_breast_cancer
data = load_breast_cancer()

df = pd.DataFrame(data.data, columns=data['feature_names'])
df['target'] = data.target
df
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# o ideal é usar diferentes métodos de correlação dependendo do tipo de dado
df_corr = df.corr(method='pearson')

plt.figure(figsize=(15, 10))

sns.heatmap(df_corr, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Heatmap da Matriz de Correlação')
plt.show()
```

```python
correlations = df.corr(method='pearson')['target'].drop('target')

# Ordena pela "força" da correlação (valor absoluto)
top_features = correlations.abs().sort_values(ascending=False)
top_features = top_features[top_features >= 0.3]
top_features
```
