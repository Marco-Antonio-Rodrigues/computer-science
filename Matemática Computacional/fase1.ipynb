```python
from mip import Model, xsum, maximize, CONTINUOUS

m = Model("Simples - Fase 1")

#Entradas

I = [1,2,3]
n = 3
S = [2,-1,1]


x = [m.add_var(var_type=CONTINUOUS, lb = 0) for i in range(n)]

m.objective = maximize(xsum((S[i]*x[i] for i in range(n))))

#s.a.
scalar = [3,1,1]
m += xsum(x[i]*scalar[i] for i in range(n)) <= 60

scalar = [1,-1,2]
m += xsum(x[i]*scalar[i] for i in range(n)) <= 10

scalar = [1,1,-1]
m += xsum(x[i]*scalar[i] for i in range(n)) <= 20

m += xsum(x[i] for i in range(n)) >= 0

m.write('model.lp')

m.optimize()

for i in range(n):
  print("Valores para: x{} = {} ".format(I[i], x[i].x ))
```
