```python
from mip import Model, minimize, INTEGER, BINARY, xsum,OptimizationStatus

model = Model(sense=minimize)

# Considerações
# 1 - a soma total de ofertas e demandas é igual a zero.

# ENTRADAS
n = 4 #nos
arcos = [(0, 1), (0, 2), (1, 2), (1, 3), (2, 3)] # lista de arcos (i, j)
u = [10, 5, 10, 5, 10] #fluxo max
c = [2, 1, 4, 3, 2] #custo por unidade de fluxo
f = [1, 1, 3, 2, 2] #custo fixo
b = [5, 0, 0, -5] #oferta e demanda do no

# VARIÁVEIS

# Variáveis de fluxo nos arcos (quanto de fluxo passa em cada arco)
x = [model.add_var(lb=0, ub=u[i], var_type=INTEGER) for i in range(len(arcos))]

# Variáveis binárias indicando se um arco está sendo usado
y = [model.add_var(var_type=BINARY) for i in range(len(arcos))]

# S.A

# Restrição de conservação de fluxo para cada nó
for i in range(n):
    fluxo_entrando = xsum(x[j] for j in range(len(arcos)) if arcos[j][1] == i)
    fluxo_saindo = xsum(x[j] for j in range(len(arcos)) if arcos[j][0] == i)
    model += fluxo_entrando - fluxo_saindo == b[i]
    print(fluxo_entrando - fluxo_saindo == b[i])

print(f"\n")
# Restrição de ativação de arcos
for i in range(len(arcos)):
    model += x[i] <= u[i] * y[i]
    print(x[i] <= u[i] * y[i])


# Objetivo
model.objective = xsum(c[i] * x[i] + f[i] * y[i] for i in range(len(arcos)))

status = model.optimize()

# Solução
if status == OptimizationStatus.OPTIMAL:
    for i in range(len(arcos)):
        print(f"Fluxo no arco {arcos[i]}: {x[i].x}")
        print(f"Arco {arcos[i]} utilizado? {'Sim' if y[i].x >= 0.99 else 'Não'}")
else:
    print("Sem solução ótima!")

```
