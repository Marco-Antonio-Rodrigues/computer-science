```python
from mip import Model, xsum, maximize, CONTINUOUS

m = Model("Exemplo Fazenda :D - Questao 1 da Lista")

#Entradas

max_ha = 200
max_est = 700000
I = ['trigo', 'arroz', 'milho']
V = [1800, 2100, 2900]
min_plant = [12, 16, 20]
L = [1.2, 0.6, 0.28]
n = 3


#variaveis: x_i = qtd (ha) platando de cada insumo i em I
x = [m.add_var(var_type=CONTINUOUS, lb = 0) for i in range(n)]


#fo: lucro é dado pelo soma total da venda de cada kilo de insumo gerado
m.objective = maximize(xsum((L[i]*V[i])*x[i] for i in range(n)))

#s.a.

#restricao de area (ha) maxima permita p/ plantacao
m += xsum(x[i] for i in range(n)) <= max_ha

#restricao de estocagem (kg) maxima permita p/ plantacao
m += xsum(V[i]*x[i] for i in range(n)) <= max_est

#restricao de consumo minimo interno da fazenda
for i in range(n):
  m += x[i] >= min_plant[i]

m.write('model.lp')

m.optimize()

for i in range(n):
  print("Ha plantado: x_{} = {} ".format(I[i], x[i].x ))

```
