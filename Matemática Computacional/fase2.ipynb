```python
from mip import Model, xsum, minimize, CONTINUOUS

m = Model("Simplex - Fase 2")

#Entradas

I = [1,2]
n = 2
S = [4,3]


x = [m.add_var(var_type=CONTINUOUS, lb = 0) for i in range(n)]

m.objective = minimize(xsum((S[i]*x[i] for i in range(n))))

#s.a.
scalar = [2,1]
m += xsum(x[i]*scalar[i] for i in range(n)) >= 10

scalar = [-3,2]
m += xsum(x[i]*scalar[i] for i in range(n)) <= 6

scalar = [1,1]
m += xsum(x[i]*scalar[i] for i in range(n)) >= 6

m += xsum(x[i] for i in range(n)) >= 0

m.write('model.lp')

m.optimize()

for i in range(n):
  print("Valores para: x{} = {} ".format(I[i], x[i].x ))
```
